```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import clear_output
```

```python
def step_function(x):
    if  x < thres:
        return 0
    else:
        return 1
```

```python
def gen_training_dat(data_point):
    x1 = np.random.random(data_point)
    x2 = np.random.random(data_point)
    y = ((x1+x2)>1).astype(int)
    
    training_set = [ ( (x1[i] , x2[i]),  y[i]  ) for i in range(len(x1)) ]
    return  training_set
 
```

```python
thres = 0.5
w = np.array([0.3, 0.9])
lr = 0.1
data_point = 100
epoch = 10
training_set = gen_training_dat(data_point)
```

```python
training_set[5:15]
```

```python
plt.figure(0)
plt.ylim(-0.1 , 1.1)
plt.xlim(-0.1 , 1.1)
ax =plt.gca()
ax.set_aspect('equal', adjustable = 'box')

for x, y in training_set :
    if y == 1 :
        plt.plot(x[0] ,x[1], 'bo')
    else:
        plt.plot(x[0], x[1], 'go')
plt.show()
    
    
```

```python
%matplotlib inline
from time import sleep

plt.figure(0)
plt.ylim(-0.1 , 1.1)
plt.xlim(-0.1 , 1.1)
ax =plt.gca() # The part I don't understand
ax.set_aspect('equal', adjustable = 'box') #The part I don't understand

xx = np.linspace(0,1,50)
for i in range(epoch):
    cnt = 0
    for x, y in training_set:
        clear_output(wait = True) #Animation effects (smoothing)
        u = sum (x*w)
        error = y - step_function(u)
        for index, value in enumerate(x):
            w[index] = w[index] + lr*error*value

        for xs, ys in training_set[0:cnt]:
            plt.ylim(-0.1, 1.1)
            plt.xlim(-0.1, 1.1)
            ax = plt.gca()
            ax.set_aspect('equal', adjustable='box')

            if ys == 1:
                plt.plot(xs[0], xs[1], 'bo')
            else:
                plt.plot(xs[0], xs[1], 'go')

        yy = -w[1] / w[0]  * xx + thres / w[0] 

        plt.plot(xx,yy)
        plt.show()
        cnt=cnt+1
        sleep(0.01) 
```
